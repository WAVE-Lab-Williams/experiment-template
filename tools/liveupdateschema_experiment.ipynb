## Updates your experiment schema while link is live!

ABSOLUTELY only run this experiment if you are 1000% sure you want to update the schema of your experiment while it is live. Make sure you've already done everything in setup_experiment.ipynb to make sure you can connect to the wave client correctly and have all dependencies.

```python
## HOW TO ENTER THE CONFIG
## Write each setting as  name=value  inside dict(...), using the same names as
## CONFIG_DEFAULTS in src/js/core/params.js. No quotes needed around the names.
##   params.js:  {number_of_repetitions: 1, base_circle_colors: ["blue","orange"]}
##   here:       dict(number_of_repetitions=1, base_circle_colors=["blue", "orange"])
##
## Remember:
##   - use  =  instead of  :
##   - JS true / false / null  ->  Python True / False / None
##   - set_config REPLACES the whole config: any setting you leave out falls back
##     to its params.js default. Include EVERY setting you want changed.
##
## check_config_keys() compares your setting names against params.js and stops
## with an error (sending nothing) if any name doesn't match, e.g. a typo.

from wave_client import WaveClient
from utils import load_environment_variables, check_config_keys

RESEARCHER_API_KEY, EXPERIMENTEE_API_KEY, WAVE_BACKEND_URL = load_environment_variables()

config = check_config_keys(dict(
    number_of_repetitions=1,
    base_circle_colors=["blue", "orange"],
))

async with WaveClient(api_key=RESEARCHER_API_KEY, base_url=WAVE_BACKEND_URL) as client:
    updated = await client.experiments.set_config(
        experiment_uuid="insert long uuid of the target experiment here", # id of whatever experiment you want to update
        config=config,
    )
    print(updated["config"])
```
